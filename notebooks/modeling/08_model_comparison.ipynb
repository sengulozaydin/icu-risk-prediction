```python
import pandas as pd

results = pd.DataFrame({
    "Model": [
        "Logistic Regression",
        "KNN",
        "Decision Tree",
        "Random Forest",
        "XGBoost",
        "Deep Learning"
    ],
    "ROC_AUC": [
        0.842,
        0.739,
        0.772,
        0.862,
        0.864,
        0.868
    ],
    "Precision": [
        0.400,
        0.236,
        0.312,
        0.455,
        0.446,
        0.404
    ],
    "Recall": [
        0.570,
        0.708,
        0.536,
        0.552,
        0.587,
        0.651
    ],
    "F1": [
        0.470,
        0.353,
        0.394,
        0.498,
        0.507,
        0.499
    ],
    "Selected_Setting": [
        "Threshold = 0.2",
        "SMOTE + K = 3",
        "Tuned + balanced + threshold = 0.6",
        "Tuned + balanced",
        "Tuned + scale_pos_weight + threshold = 0.6",
        "Class weight + threshold = 0.6"
    ]
})

results
```

```python
import matplotlib.pyplot as plt

plot_data = results.set_index("Model")[["Precision", "Recall", "F1"]]

plot_data.plot(
    kind="bar",
    figsize=(12, 6)
)

plt.title("Model Comparison: Precision, Recall and F1")
plt.ylabel("Score")
plt.ylim(0, 1)
plt.xticks(rotation=45, ha="right")
plt.legend(title="Metric")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 5))

plt.bar(
    results["Model"],
    results["ROC_AUC"]
)

plt.title("Model Comparison: ROC-AUC")
plt.ylabel("ROC-AUC")
plt.ylim(0, 1)
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
final_models = results[
    results["Model"].isin([
        "Random Forest",
        "XGBoost",
        "Deep Learning"
    ])
].set_index("Model")[["ROC_AUC", "Precision", "Recall", "F1"]]

final_models.plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title("Final Model Comparison")
plt.ylabel("Score")
plt.ylim(0, 1)
plt.xticks(rotation=0)
plt.legend(title="Metric")
plt.tight_layout()
plt.show()
```

### Final Model Candidates

Random Forest, XGBoost and Deep Learning were the strongest models overall.

Among them, Deep Learning was selected as the final candidate because recall was the main priority for ICU mortality prediction.

- Random Forest: Precision **0.455**, Recall **0.552**
- XGBoost: Precision **0.446**, Recall **0.587**
- Deep Learning: Precision **0.404**, Recall **0.651**

Deep Learning provided the highest recall while keeping precision at a reasonable level.
